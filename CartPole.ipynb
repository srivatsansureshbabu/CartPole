```python
import gymnasium as gym
from stable_baselines3 import SAC
import matplotlib.pyplot as plt
from IPython import display
import numpy as np
import imageio
import cv2
from stable_baselines3 import PPO
```

```python
env = gym.make('CartPole-v1', render_mode="rgb_array")
```

```python
import warnings

warnings.filterwarnings("ignore", category=UserWarning, module="imageio")
```

```python
def policy(observation,prev_action):
    cart_position, cart_velocity, pole_angle, pole_angular_velocity = observation

    if( abs(pole_angle) > 0.100 ):
      if( pole_angle < 0):
        return 1
      else:
        return 0
    else:
      if(prev_action is None):
        return 1
      else:
        return not prev_action
    # return cart_position,cart_velocity,pole_angle,
```

```python
model = PPO("MlpPolicy",env,verbose=1)
model.learn(total_timesteps=100000)

obs = env.reset()

```

```python
# observation = env.reset()
# frame = np.array(env.render())
# print(frame.shape)
# frame = frame.squeeze()
# print(frame.shape)
# plt.imshow(frame)
# print(np.array(frame))
frames = []
observation,info = env.reset()
prev_action = 0
for _ in range(1000):
    # print(f"Step: {_/500}", end="\r")
    frame = np.array(env.render())
    frame = frame.squeeze()
    frames.append(frame.astype(np.uint8))


    # action = env.action_space.sample()
    # print("typeAction: ", type(action))
    cart_position, cart_velocity, pole_angle, pole_angular_velocity = observation
    action,_ = model.predict(observation)
    observation, reward, terminated, truncated,info = env.step(action)
    prev_action = action
    print("Reward: ",reward)


    print("__FRAME: ",_)
    print(f"θ={pole_angle:+.3f} | θ_dot={pole_angular_velocity:+.3f} | action={action}",flush=True)



    # imageio.mimsave("cartpole.mp4", frames, fps=30,macro_block_size=1)
    if terminated or truncated:
        print("FAILED!!!!")
        observation,_ = env.reset()
imageio.mimsave("CartPole_Balancing_RL_100k.mp4", frames, fps=30,macro_block_size=1)

env.close()
```
